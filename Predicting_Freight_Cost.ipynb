```python
vendor_df = pd.read_sql_query(f"select * from vendor_invoice",conn)
vendor_df.head()
```

```python
vendor_df[["Quantity","Freight","Dollars"]].corr()
```

```python
# Relationship between Quantity, Dollars and Freight

plt.figure(figsize=(4, 2))

sns.heatmap(
    vendor_df[['Quantity', 'Dollars', 'Freight']].corr(),
    annot=True
)

plt.show()


# Scatter plots
plt.figure(figsize=(7, 5))

plt.scatter(
    vendor_df["Quantity"],
    vendor_df["Freight"],
    color="#f57a55"
)

plt.scatter(
    vendor_df["Dollars"],
    vendor_df["Freight"],
    color="#7fe55a"
)

plt.legend(["Quantity", "Dollars"])
plt.xlabel("Quantity / Dollars")
plt.ylabel("Freight Cost")
plt.title("Relationship between Quantity, Dollars and Freight")

plt.show()
```
